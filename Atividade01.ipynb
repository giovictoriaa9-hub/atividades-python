```
# Questão 6
numeros = []
while len(numeros) < 3:
    try:
        n = int(input("Digite um número inteiro: "))
        numeros.append(n)
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
maior = numeros[0]
for n in numeros:
    if n > maior:
        maior = n
print("Maior número:", maior)
```

```
# Questão 10
valor = 0
while valor <= 0:
    try:
        valor = int(input("Digite um valor positivo: "))
        if valor <= 0:
            print("O valor deve ser positivo!")
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
cedulas = (100, 50, 20, 10, 5, 2, 1)
total = 0
for cedula in cedulas:
    quantidade = valor // cedula
    valor = valor % cedula
    print("Cédulas de R$", cedula, ":", quantidade)
    total += quantidade
print("Total de cédulas:", total)
```

```
# Questão 11
n = 0
while n <= 1:
    try:
        n = int(input("Digite um inteiro maior que 1: "))
        if n <= 1:
            print("O número deve ser maior que 1!")
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
divisores = 0
for i in range(1, n + 1):
    if n % i == 0:
        divisores += 1
if divisores == 2:
    print("O número é primo!")
else:
    print("O número não é primo!")
```

```
# Questão 13
a = 0
b = 0
while a <= 0:
    try:
        a = int(input("Digite A: "))
        if a <= 0:
            print("A deve ser positivo!")
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
while b <= a:
    try:
        b = int(input("Digite B: "))
        if b <= a:
            print("B deve ser maior que A!")
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
quantidade = 0
print("Números primos encontrados:")
for n in range(a, b + 1):
    divisores = 0
    for i in range(1, n + 1):
        if n % i == 0:
            divisores += 1
    if divisores == 2:
        print(n)
        quantidade += 1
print("Quantidade de primos:", quantidade)
```

```
# Questão 14
secreto = 37
tentativas = 0
palpite = 0
while tentativas < 5 and palpite != secreto:
    try:
        palpite = int(input("Digite seu palpite: "))
        tentativas += 1
        if palpite == secreto:
            print("Parabéns! Você acertou!")
            print("Tentativas utilizadas:", tentativas)
        elif palpite < secreto:
            print("O número secreto é maior!")
        else:
            print("O número secreto é menor!")
    except ValueError:
        print("Entrada inválida! Digite um inteiro.")
if palpite != secreto:
    print("Suas tentativas acabaram!")
    print("O número secreto era:", secreto)
```

```
# Questão 15
saldo = 200
entradas = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0
codigo = -1
while codigo != 0:
    try:
        codigo = int(input("Digite o código (1-Entrada, 2-Retirada, 0-Sair): "))
        if codigo == 1:
            valor = 0
            while valor <= 0:
                try:
                    valor = float(input("Digite o valor da entrada: "))
                    if valor <= 0:
                        print("O valor deve ser positivo!")
                except ValueError:
                    print("Digite um valor numérico!")
            saldo += valor
            entradas += 1
            total_entrada += valor
            print("Entrada realizada!")
        elif codigo == 2:
            valor = 0
            while valor <= 0:
                try:
                    valor = float(input("Digite o valor da retirada: "))
                    if valor <= 0:
                        print("O valor deve ser positivo!")
                except ValueError:
                    print("Digite um valor numérico!")
            if valor <= saldo:
                saldo -= valor
                retiradas += 1
                total_saida += valor
                print("Retirada realizada!")
            else:
                recusadas += 1
                print("Retirada recusada! Saldo insuficiente.")
        elif codigo == 0:
            print("Encerrando o caixa...")
        else:
            print("Código inválido!")
    except ValueError:
        print("Digite apenas números inteiros!")
print("\nRESUMO DO CAIXA")
print("Saldo final: R$", saldo)
print("Quantidade de entradas:", entradas)
print("Quantidade de retiradas:", retiradas)
print("Retiradas recusadas:", recusadas)
print("Total que entrou: R$", total_entrada)
print("Total que saiu: R$", total_saida)
```
