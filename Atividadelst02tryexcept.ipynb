```python
# Questão 4
horas = 0
while horas <= 0:
    try:
        horas = int(input("Digite a quantidade de horas: "))
        if horas <= 0:
            print("Digite uma quantidade positiva!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
if horas <= 2:
    valor = 8
elif horas <= 5:
    valor = 8 + (horas - 2) * 3
else:
    valor = 17 + (horas - 5) * 2
print(f"Valor a pagar: R$ {valor:.2f}")
```

```python
# Questão 5
preco = 25
total = 0
while total < preco:
    try:
        valor = int(input("Digite o valor a inserir: R$ "))
        if valor <= 0:
            print("Digite um valor maior que zero!")
        else:
            total = total + valor
            if total < preco:
                print("Faltam R$", preco - total)
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
troco = total - preco
print("Total inserido: R$", total)
print("Troco: R$", troco)
```

```python
# Questão 6
contador = 1
maior = 0
posicao = 0
while contador <= 8:
    try:
        numero = int(input(f"Digite o {contador}º número: "))
        if contador == 1:
            maior = numero
            posicao = contador
        elif numero > maior:
            maior = numero
            posicao = contador
        contador = contador + 1
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
print("Maior valor:", maior)
print("Primeira posição:", posicao)
```

```python
# Questão 7
contador = 1
maior = 0
segundo = 0
while contador <= 8:
    try:
        numero = int(input(f"Digite o {contador}º número: "))
        if contador == 1:
            maior = numero
            contador = contador + 1
        elif contador == 2:
            if numero == maior:
                print("Digite um número diferente!")
            else:
                if numero > maior:
                    segundo = maior
                    maior = numero
                else:
                    segundo = numero
                contador = contador + 1
        else:
            if numero == maior or numero == segundo:
                print("Digite um número diferente!")
            elif numero > maior:
                segundo = maior
                maior = numero
                contador = contador + 1
            elif numero > segundo:
                segundo = numero
                contador = contador + 1
            else:
                contador = contador + 1
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
print("Maior valor:", maior)
print("Segundo maior valor:", segundo)
```

```python
# Questão 8
contador = 1
anterior = 0
repetiu = False
while contador <= 8:
    try:
        numero = int(input(f"Digite o {contador}º número: "))
        if contador > 1:
            if numero == anterior:
                repetiu = True
        anterior = numero
        contador = contador + 1
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
# Questão 9
saldo = 500
contador = 1
while contador <= 6:
    print(f"\nOperação {contador}")
    tipo = 0
    while tipo != 1 and tipo != 2:
        try:
            tipo = int(input("1 - Depósito / 2 - Saque: "))
            if tipo != 1 and tipo != 2:
                print("Opção inválida! Digite 1 ou 2.")
        except ValueError:
            print("Erro! Digite apenas números inteiros.")
    valor = 0
    while valor <= 0:
        try:
            valor = float(input("Digite o valor: R$ "))
            if valor <= 0:
                print("Digite um valor maior que zero!")
        except ValueError:
            print("Erro! Digite um valor numérico válido.")
    if tipo == 1:
        saldo = saldo + valor
        print("Depósito realizado!")
    elif tipo == 2:
        if valor <= saldo:
            saldo = saldo - valor
            print("Saque realizado!")
        else:
            print("Saldo insuficiente.")
    contador = contador + 1
print(f"\nSaldo final: R$ {saldo:.2f}")
```

```python
# Questão 10
valor = 0
while valor <= 0:
    try:
        valor = int(input("Digite o valor: R$ "))
        if valor <= 0:
            print("Digite um valor inteiro positivo!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
resto = valor
ced100 = resto // 100
resto = resto % 100
ced50 = resto // 50
resto = resto % 50
ced20 = resto // 20
resto = resto % 20
ced10 = resto // 10
resto = resto % 10
ced5 = resto // 5
resto = resto % 5
ced2 = resto // 2
resto = resto % 2
ced1 = resto
total = ced100 + ced50 + ced20 + ced10 + ced5 + ced2 + ced1
print("\nQuantidade de cédulas:")
print("R$ 100:", ced100)
print("R$ 50:", ced50)
print("R$ 20:", ced20)
print("R$ 10:", ced10)
print("R$ 5:", ced5)
print("R$ 2:", ced2)
print("R$ 1:", ced1)
print("Total de cédulas:", total)
```

```python
# Questão 12
n = 0
while n <= 0:
    try:
        n = int(input("Digite um número inteiro positivo: "))
        if n <= 0:
            print("Digite um número maior que zero!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
divisor = 1
quantidade = 0
print("Divisores de", n, ":")
while divisor <= n:
    if n % divisor == 0:
        print(divisor)
        quantidade = quantidade + 1
    divisor = divisor + 1
print("Quantidade de divisores:", quantidade)
```

```python
# Questão 13
a = 0
b = 0
while a <= 0:
    try:
        a = int(input("Digite o valor de A: "))
        if a <= 0:
            print("A deve ser positivo!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
while b <= a:
    try:
        b = int(input("Digite o valor de B: "))
        if b <= 0:
            print("B deve ser positivo!")
        elif b <= a:
            print("B deve ser maior que A!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
quantidade = 0
numero = a
print("Números primos encontrados:")
while numero <= b:
    if numero > 1:
        divisor = 2
        divisores = 0
        while divisor < numero:
            if numero % divisor == 0:
                divisores = divisores + 1
            divisor = divisor + 1
        if divisores == 0:
            print(numero)
            quantidade = quantidade + 1
    numero = numero + 1
print("Quantidade de primos:", quantidade)
```

```python
# Questão 14
secreto = 37
tentativas = 0
palpite = 0
while tentativas < 5 and palpite != secreto:
    try:
        palpite = int(input("Digite seu palpite: "))
        tentativas = tentativas + 1
        if palpite == secreto:
            print("Parabéns! Você acertou!")
            print("Tentativas utilizadas:", tentativas)
        elif palpite < secreto:
            print("O número secreto é maior!")
        else:
            print("O número secreto é menor!")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
if palpite != secreto:
    print("Suas 5 tentativas acabaram!")
    print("O número secreto era:", secreto)
```

```python
# Questão 15
saldo = 200
codigo = -1
entradas = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0
while codigo != 0:
    try:
        print("\n1 - Entrada")
        print("2 - Retirada")
        print("0 - Encerrar")
        codigo = int(input("Digite o código: "))
        if codigo == 0:
            print("Caixa encerrado!")
        elif codigo == 1 or codigo == 2:
           valor = 0
        while valor <= 0:
                try:
                    valor = float(input("Digite o valor: R$ "))
                    if valor <= 0:
                        print("Digite um valor maior que zero!")
                except ValueError:
                    print("Erro! Digite um número válido.")
        if codigo == 1:
                saldo = saldo + valor
                total_entrada = total_entrada + valor
                entradas = entradas + 1
                print("Entrada realizada!")
        elif codigo == 2:
                if valor <= saldo:
                    saldo = saldo - valor
                    total_saida = total_saida + valor
                    retiradas = retiradas + 1
                    print("Retirada realizada!")
                else:
                    recusadas = recusadas + 1
                    print("Saldo insuficiente!")
        else:
            print("Código inválido! Digite 0, 1 ou 2.")
    except ValueError:
        print("Erro! Digite apenas números inteiros.")
print("\nRESUMO DO CAIXA")
print(f"Saldo final: R$ {saldo:.2f}")
print("Quantidade de entradas:", entradas)
print("Retiradas realizadas:", retiradas)
print("Retiradas recusadas:", recusadas)
print(f"Total que entrou: R$ {total_entrada:.2f}")
print(f"Total que saiu: R$ {total_saida:.2f}")
```
