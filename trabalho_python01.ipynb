```python
# Questão 05
nota1= float(input("Digite a nota: "))
nota2= float(input("Digite a nota: "))
media = (nota1 + nota2) / 2
if 7 <= media <= 10:
 print("Aprovado")
elif media >= 5:
 print("Recuperação")
elif 0 <= media < 5:
 print("Reprovado")
else:
  print("Nota inválida")
```

```python
# Questão 14
num_secreto = 7
tentativas = 0
n1 = int(input("Adivinhe o número: "))
while n1 != num_secreto:
  tentativas += 1
  if n1 > num_secreto:
   print("Tente um número menor")
  else:
   print("Tente um número maior")
  n1 = int(input("Adivinhe o número novamente: "))
tentativas += 1
print("Parabéns! Você acertou")
print("Número de tentativas:", tentativas)
```

```python
# Questão 15
quant_digitos = 0
positivos = 0
negativos = 0
soma_digitos = 0
n1 = int(input("Digite um número: "))
while n1 != 0:
  quant_digitos += 1
  soma_digitos += n1
  if n1 > 0:
    positivos += 1
  else:
    negativos +=1
  n1 = int(input("Digite um número: "))
print("Quantidade de dígitos:", quant_digitos)
print("Soma dos dígitos:", soma_digitos)
print("Quantidade de números positivos:", positivos)
print("Quantidade de números negativos:", negativos)
```
