```python
# Questão 4
estacionamento = int(input("Quantidade de horas: "))
while estacionamento < 0:
 print("Horas inválidas")
 estacionamento = int(input("Quantidade de horas: "))
 if estacionamento >= 0:
  break
if estacionamento <= 2:
 print("Valor a pagar: R$8,00")
elif 2 < estacionamento <= 5:
 valor = 8 + (estacionamento - 2) * 2
 print(f"Valor a pagar: R${valor}")
elif estacionamento > 5:
  valor = 17 + (estacionamento - 5) * 3
  print(f"Valor a pagar: R${valor}")
```

```python
# Questão 9
saldo = 500
for i in range(6):
  pergunta = int(input("Escolha:\n1 - depósito \n2 - saque\nDigite um desses números aqui: "))
  while pergunta != 1 and pergunta != 2:
     print("Dígito inválido.")
     pergunta = int(input("Escolha:\n1 - depósito\n2 - saque\nDigite um desses números aqui: "))
  if pergunta == 1:
    deposito = int(input("O quanto você quer depositar? "))
    saldo += deposito
    print(f"Saldo atual: {saldo}")
  elif pergunta == 2:
    saque = int(input('O quanto você quer sacar? '))
    if saque > saldo:
       print("Saldo insuficiente para saque.")
    else:
      saldo -= saque
      print(f"Saque concluído com sucesso!")
      print(f"Saldo atual: {saldo}")
```

```python
# Questão 15
saldo = 200
entradas = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0
codigo = -1
while codigo != 0:
  codigo = int(input("Escolha: \n1=entrada\n2=retirada\n0=sair\nDigite a numeração que corresponde à sua decisão: "))
  if codigo == 1:
      valor = float(input("Valor de entrada: "))
      saldo += valor
      entradas += 1
      total_entrada += valor
  elif codigo == 2:
   valor = float(input("Valor de retirada: "))
   if valor <= saldo:
    saldo -= valor
    retiradas += 1
    total_saida += valor
   else:
    print("Saldo insuficiente")
    recusadas += 1
  elif codigo != 0:
   print("Código inválido")
print("Saldo final:", saldo)
print("Entradas:", entradas)
print("Retiradas:", retiradas)
print("Retiradas recusadas:", recusadas)
print("Total que entrou:", total_entrada)
print("Total que saiu:", total_saida)
```
