```python
# Questão 06
nomes = []
nome = input("Digite um nome (ou FIM para parar): ")
while nome != "FIM":
  contador = 0
  for n in nomes:
   if n == nome:
    contador + 1
   if contador == 0:
    nomes.append(nome)
  nome = input("Digite um nome (ou FIM para parar): ")
print("Lista:", nomes)
print("Quantidade de nomes diferentes:", len(nomes))
```

```python
# Questão 07
semana = { 1: "Domingo", 2: "Segunda-feira", 3: "Terça-feira", 4: "Quarta-feira", 5: "Quinta-feira", 6: "Sexta-feira", 7: "Sábado"}
dia = -1
while dia != 0:
 dia = int(input("Digite um número de 1 a 7 (ou 0 para sair): "))
 if dia == 0:
  print("Programa encerrado.")
 elif dia < 1 or dia > 7:
  print("Dia da semana não existente")
 else:
  print("O dia da semana é:", semana[dia])
```

```python
# Questão 11
contagem = {}
for i in range(8):
  palavra = input("Digite uma palavra: ")
  if palavra in contagem:
    contagem[palavra] = contagem[palavra] + 1
  else:
    contagem[palavra] = 1
print("\nResultado da contagem:")
for palavra in contagem:
    print(palavra, "apareceu", contagem[palavra], "vez(es)")
```

```python
# Questão 12
estoque = {"Caderno": 50, "Caneta": 100, "Lápis": 80, "Borracha": 30}
produto_vendido = input("Digite o nome do produto vendido: ")
quantidade_venda = int(input("Digite a quantidade vendida: "))
if produto_vendido in estoque:
    if estoque[produto_vendido] >= quantidade_venda:
      estoque[produto_vendido] = estoque[produto_vendido] - quantidade_venda
      print("Venda realizada com sucesso!")
    else:
      print("Erro: Estoque insuficiente!")
else:
  print("Erro: Produto não cadastrado!")
print("\nEstoque final:")
for produto in estoque:
    print(produto, ":", estoque[produto], "unidades")
```

```python
# Questão 13
lista_alunos = []
for i in range(5):
    print("--- Cadastro do Aluno", i + 1, "---")
    nome = input("Digite o nome do aluno: ")
    nota1 = float(input("Digite a primeira nota: "))
    nota2 = float(input("Digite a segunda nota: "))
    media = (nota1 + nota2) / 2
    aluno = {"nome": nome, "nota1": nota1, "nota2": nota2, "media": media}
    lista_alunos.append(aluno)
print("\n--- Relatório Final (Nome e Média) ---")
for aluno in lista_alunos:
    print("Aluno:", aluno["nome"], "/ Média:", aluno["media"])
```

```python
# Questão 14
lista_produtos = []
for i in range(5):
  print("--- Cadastro do Produto", i + 1, "---")
  codigo = int(input("Digite o código do produto: "))
  nome = input("Digite o nome do produto: ")
  preco = float(input("Digite o preço do produto: "))
  produto = (codigo, nome, preco)
  lista_produtos.append(produto)
print("\n--- Consulta de Produto ---")
codigo_busca = int(input("Digite o código do produto para buscar: "))
contador = 0
for prod in lista_produtos:
  if prod[0] == codigo_busca:
    print("Produto encontrado!")
    print("Nome:", prod[1])
    print("Preço: R$", prod[2])
    contador = contador + 1
if contador == 0:
    print("Código não encontrado!")
```

```python
# Questão 15
estoque_produtos = []
for i in range(5):
  print("--- Cadastro do Item", i + 1, "---")
  codigo = int(input("Digite o código: "))
  nome = input("Digite o nome: ")
  quantidade = int(input("Digite a quantidade: "))
  preco = float(input("Digite o preço: "))
  item = {"codigo": codigo, "nome": nome, "quantidade": quantidade, "preco": preco}
  estoque_produtos.append(item)
print("\n--- Todos os Itens Cadastrados ---")
for item in estoque_produtos:
  print("Código:", item["codigo"], "/ Nome:", item["nome"], "/ Qtd:", item["quantidade"], "/ Preço: R$", item["preco"])
valor_total_estoque = 0.0
maior_valor_acumulado = 0.0
nome_item_maior_valor = ""
for item in estoque_produtos:
  valor_item_atual = item["quantidade"] * item["preco"]
  valor_total_estoque = valor_total_estoque + valor_item_atual
  if valor_item_atual > maior_valor_acumulado:
    maior_valor_acumulado = valor_item_atual
    nome_item_maior_valor = item["nome"]
print("\n--- Resultados das Análises ---")
print("Valor total de todo o estoque: R$", valor_total_estoque)
print("Item com maior valor em estoque:", nome_item_maior_valor, "(Total: R$", maior_valor_acumulado, ")")
quantidade_cadastrada = len(estoque_produtos)
tupla_resumo = (quantidade_cadastrada, valor_total_estoque)
print("Tupla Resumo gerada:", tupla_resumo)
```
